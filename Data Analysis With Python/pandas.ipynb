```python
## Series -> Pandas series is a one dimensional array like object that can hold any data type, It is similar to a column in a table

import pandas as pd

data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
print(type(series))
```

```python
## Series from dictionary 

data={'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index=['a','b','c']

pd.Series(data,index=index)

```

```python
#Dataframe -> Tabular form
#Create a dataframe from a dictionary of list

data={
    'Name':['Abhay','Raj','John'],
    'Age':[25,30,45],
    'City':['Bangalore','New York','Texas']
}

df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a dataframe from a list of dictionaries

data=[
   {'Name':'Abhay','Age':32,'City':'Bangalore'},
    {'Name':'Raj','Age':21,'City':'Gkp'},
    {'Name':'Bappy','Age':32,'City':'Jhk'},
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
