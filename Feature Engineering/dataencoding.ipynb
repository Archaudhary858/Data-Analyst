Data Encoding

Nominal/OHE Encoding

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## Create a simple dataframe

df=pd.DataFrame({
    'color': ['red','blue','green','red','blue']
})
```

```python
df.head()
```

Converting Categorical features into numerical features

```python
##create a instance of onehotencoder : We need OneHotEncoder() to create a reusable tool that automatically converts categorical values into numbers in a way ML models can understand.
encoder=OneHotEncoder()
```

```python
##Perform fit and transform
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python

encoder_df 
```

```python
# For new Data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

Ordinal Encoding

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
```

```python
df2= pd.DataFrame({
    'size' : ['small','medium','large','medium','small','large']
})
```

```python
df2
```

```python
## Create  a instance of ordinal encoder and then fit_transform
encoder = OrdinalEncoder(categories=[['small','medium','large']])

## Categories used to give ranks to the data we want to transform
```

```python
encoder.fit_transform(df2[['size']])
```

```python
## Giving new data
encoder.transform([['small']])
```

Target Guided Ordinal Encoding

```python
df3 = pd.DataFrame({
    'city': ['New York','London','Paris','Tokyo','New York','Paris'],
    'price': [200,150,300,250,180,320]
})
```

```python
df3
```

```python
mean_pricce= df3.groupby('city')['price'].mean().to_dict()
```

```python
mean_pricce
```

```python
df3['city_encoded'] = df3['city'].map(mean_pricce)
```

```python
df3
```

```python
df3[['price','city_encoded']]
```
