Handling Imbalanced Dataset

1.Up Sampling
2.Down Sampling

```python
import numpy as np
import pandas as pd

#Set the random seed for reproductibility
np.random.seed(123)

#Create a datafram with two classes

n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio)
n_class_1=n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python

## CREATE MY DATAFRAME WITH IMBALANCED DATASET
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
df.tail()
```

```python
df['target'].value_counts()
```

```python
## Upsampling

df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_upsample = resample(
         df_minority,
         replace=True,
         n_samples=len(df_majority),
         random_state=42)

'''replace=True → allows the same row to be picked multiple times (sampling with replacement).

      Without this, you couldn’t make the minority as big as the majority, since you’d run out of rows.

n_samples=len(df_majority) → makes the minority class as big as the majority class.

      Example: if majority has 900 rows, this will create 900 minority rows (by repeating them).

random_state=42 → ensures reproducibility (same resampled rows every time you run).'''
```

```python
df_minority_upsample.shape
```

```python
df_minority_upsample.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsample])
```

```python
df_upsampled['target'].value_counts()
```

```python
## Down Sampling

import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df = pd.concat([class_0, class_1]).reset_index(drop=True)

# Check the class distribution
print(df['target'].value_counts())
```

```python
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_majority_downsampled=resample(df_majority,replace=False, #Sample With replacement
         n_samples=len(df_minority),
         random_state=42
        )
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minority,df_majority_downsampled])
df_downsampled.target.value_counts()
```
