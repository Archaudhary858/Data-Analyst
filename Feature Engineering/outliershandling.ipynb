5 Number Summary And Box Plot

```python
import numpy as np
```

```python
lst_marks=[45,32,34,67,39,56,78,65,99,80]
minimum,Q1,median,Q3,maximum = np.quantile(lst_marks,[0,0.25,0.5,0.75,1.0])
```

```python
minimum,Q1,maximum
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*(IQR)
higher_fence=Q3+1.5*IQR
```

```python
lower_fence
higher_fence
```
